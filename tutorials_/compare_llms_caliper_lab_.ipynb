```python
#@title 🤖🧠🇦🇮👾 >>>>> INITIALIZE OPENAI LLM API
# !pip install openai

from openai import OpenAI


def generate_openAI_response_(the_input_prompt_):
  client = OpenAI(api_key=OPENAI_API_KEY)
  response = client.responses.create(
    model="gpt-5.4-mini",
    input=the_input_prompt_,
    store=True,
  )
  return response.output_text

promptAI = "write a haiku about ai"
openAI_response = generate_openAI_response_(the_input_prompt_=promptAI)

print(openAI_response);
```

```python
#@title 🤖🧠🇦🇮👾 >>>>> INITIALIZE GEMINI LLM API

#!pip install -q -U google-generativeai

from google import genai

# Initialize the client with your API key
def generate_gemini_response_(the_input_prompt_):
  client = genai.Client(api_key=GEMINI_API_KEY)
  response = client.models.generate_content(
      # model="gemini-3.1-flash-lite-preview",
      # model="gemini-2.5-flash-lite",
      model="gemini-2.5-flash",
      contents=the_input_prompt_,
  )
  return response.text

promptAI = "Explain how AI works in a few words"
gemini_response = generate_gemini_response_(the_input_prompt_=promptAI)

print(gemini_response);
```

```python
#@title 🤖🧠🇦🇮👾 >>>>> INITIALIZE LLAMA LLM API

import requests

def generate_openrouter_response_(the_input_prompt_):
  headers = {
  'Authorization': 'Bearer '+ OPENROUTER_API_KEY}
  response = requests.post('https://openrouter.ai/api/v1/chat/completions', headers=headers, json={
    'model': 'meta-llama/llama-3.3-70b-instruct',
    'messages': [{ 'role': 'user', 'content': the_input_prompt_ }],
    'provider': {
      'sort': 'throughput',
    },
  })
  return response.json()

promptAI = "write a short para about ai, not more than 4 sentences"
openrouter_response = generate_openrouter_response_(the_input_prompt_=promptAI)

# print(openrouter_response)
more_data_ = openrouter_response['choices'][0]
print(more_data_['message']['content'])

```

```python
#@title 🛠🛠 FUNCTION DEFINITION FOR GETTING RESPONSE FROM ALL MODELS
import time
def get_response_from_llms_(prompt_qn, prmopt_ans):
  the_qn_dict_ = {}
  openAI_response = generate_openAI_response_(the_input_prompt_=prompt_qn)
  time.sleep(5)
  gemini_response = generate_gemini_response_(the_input_prompt_=prompt_qn)
  time.sleep(5)
  openrouter_op = generate_openrouter_response_(the_input_prompt_=prompt_qn)
  op_data_ = openrouter_op['choices'][0]
  openrouter_response = op_data_['message']['content']
  the_qn_dict_['QUESTION'] = prompt_qn
  the_qn_dict_['HUMAN_ANS'] = prmopt_ans
  the_qn_dict_['GEMINI_RESPONSE'] = gemini_response
  the_qn_dict_['OPENAI_RESPONSE'] = openAI_response
  the_qn_dict_['LLAMA_RESPONSE'] = openrouter_response
  return the_qn_dict_



qn_llm_response_dict_ = get_response_from_llms_(the_question_bank_['instructions_'] + the_question_bank_['question10_'], the_question_bank_['answer10_'])


print(qn_llm_response_dict_)
```

```python
#@title 📥 EVALUATION OF LLM OUTPUT (🕵 OPENAI as a JUDGE) -->> THE EVALUATION SCORE IS COMPUTED BETWEEN HUMAN RESPONSE AND AI GENERATED RESPONSE, AND THE RANGE OF THIS SCORE IS FROM 0 TO 5, WHERE A SCORE OF 4 OR ABOVE IS ACCEPTABLE

import openai
import json

# from google.colab import userdata
# KEY_OPENAI_ = userdata.get('OPENAI_ANOTHER')
# print(KEY_OPENAI_)

client = openai.OpenAI(api_key=KEY_OPENAI_)
def llm_judge(question, model_response, reference_answer=None):
    prompt = f"""
    You are an impartial judge evaluating the quality of an AI response.

    [User Question]: {question}
    [Model Response]: {model_response}
    {"[Reference Answer]: " + reference_answer if reference_answer else ""}

    Evaluate the response based on:
    1. Accuracy: Is the information correct?
    2. Completeness: Does it answer all parts of the question?

    Provide your evaluation in the following JSON format:
    {{
        "reasoning": "A brief explanation of your logic",
        "score": (A scale of 1-5, where 1 is poor and 5 is excellent)
    }}
    """

    response = client.chat.completions.create(
        model="gpt-5.4-mini-high",
        messages=[{"role": "system", "content": "You are a strict but fair evaluator."},
                  {"role": "user", "content": prompt}],
        response_format={"type": "json_object"}
    )

    return json.loads(response.choices[0].message.content)

# Example Usage
llm_data_ = qn_llm_data[2]
result = llm_judge(
    question = llm_data_['QUESTION'],
    model_response = llm_data_['OPENAI_RESPONSE'],
    reference_answer = llm_data_['HUMAN_ANS']
)

print(f"Performance: {result['score']}/5")
if result['score'] < 4:
  print(f"INCORRECT")
else:
  print(f"CORRECT")
print(f"Reasoning: {result['reasoning']}")
```

```python
#@title EVALUATE LLM AS A JUDGE AND SAVE SCORES, SCORE > 3 , ANSWER : CORRECT , ELSE, ANSWER : WRONG

for sx_ in qn_llm_data:
  print(10*'>><<>><<')
  print(sx_['QUESTION'],'\n\n')
  print(sx_['HUMAN_ANS'],'\n\n')
  ##------------##-----------##
  result = llm_judge(
    question = sx_['QUESTION'],
    model_response = sx_['OPENAI_RESPONSE'],
    reference_answer = sx_['HUMAN_ANS']
  )
  score = result['score']
  reason = result['reasoning']
  sx_['gpt-5.4-mini_llm_judge_score_OPENAI'] = score
  sx_['gpt-5.4-mini_llm_judge_reason_OPENAI'] = reason

  time.sleep(5)
  ##------------##-----------##
  result = llm_judge(
    question = sx_['QUESTION'],
    model_response = sx_['GEMINI_RESPONSE'],
    reference_answer = sx_['HUMAN_ANS']
  )
  score = result['score']
  reason = result['reasoning']
  sx_['gpt-5.4-mini_llm_judge_score_GEMINI'] = score
  sx_['gpt-5.4-mini_llm_judge_reason_GEMINI'] = reason

  time.sleep(5)
  ##------------##-----------##
  result = llm_judge(
    question = sx_['QUESTION'],
    model_response = sx_['LLAMA_RESPONSE'],
    reference_answer = sx_['HUMAN_ANS']
  )
  score = result['score']
  reason = result['reasoning']
  sx_['gpt-5.4-mini_llm_judge_score_LLAMA'] = score
  sx_['gpt-5.4-mini_llm_judge_reason_LLAMA'] = reason

```
